```python
# Comparison Operators

# equal ==
# not equal !=
# greater >
# less <
# greater than equal to >=
# less than equal to <=

x = input('Put the first value :')
y = input('Put the second value :')

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

```python
# Logical Operators

# and --> append two conditions
# or --> any of the condition to be true
# not --> (revert) True -> False | False -> True

age = 20
salary = 50000

print(age >=20 and salary > 40000)

print(age > 20 or salary > 40000)

print(not(age >= 20))
```

```python
# Conditional : if, else, elif

age = int(input('Please enter your age: '))    # voting eligibility

if age >= 18:
  print("You are eligible for voting")
elif age == 17:
   print("You are  not eligible for voting, but one year short")
else:
  print("You are not eligible for voting")

```

```python
## Classification

# problem statement : an item more than of 2000, you will get 20% discount and if your spending is beyond 6000, then they'll give you 30% discount
# 18% GST applicable

# billing system

Spending_amount = float(input('Please enter the amount you spent: '))   # enter how much spending you have done

# calculate payable amount

if Spending_amount > 2000:
  Spending_amount = Spending_amount * 0.8
elif Spending_amount > 6000:
  Spending_amount = Spending_amount * 0.7
else:
  Spending_amount = Spending_amount

# add tax amount

Spending_amount = Spending_amount * 1.18

print('Final Payable amount is:',Spending_amount)


```
